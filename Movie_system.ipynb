```python
def convert(objct):

    L = []
    for i in ast.literal_eval(objct):
        L.append(i["name"])
        return L
```

```python
def convert3(objct):

    L = []
    counter = 0
    for i in ast.literal_eval(objct):
        if counter != 0:
            3
        L.append(i["name"])
        counter += 1
        return L
```

```python
contentdataFrame["genres"] = contentdataFrame["genres"].apply(convert)
contentdataFrame["keywords"] = contentdataFrame["keywords"].apply(convert)
```

```python
import ast


def Find_Director(objct):
    L = []
    # If the column was already converted or isn't a string, return it as-is
    if isinstance(objct, list):
        return objct

    if not isinstance(objct, str) or "[" not in objct:
        return L

    try:
        for i in ast.literal_eval(objct):
            if isinstance(i, dict) and i.get("job") == "Director":
                L.append(i["name"])
                break
    except (ValueError, SyntaxError):
        pass

    return L


# CRITICAL CHANGE: Save to a NEW column named "director" so "crew" stays untouched
contentdataFrame["director"] = contentdataFrame["crew"].apply(Find_Director)

# View your results
print(contentdataFrame[["crew", "director"]].head())
```
